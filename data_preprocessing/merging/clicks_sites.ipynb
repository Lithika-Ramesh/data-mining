# Clicks and Sites Integration & Aggregation


## 1. Load Source Data


```python
import pandas as pd
from pathlib import Path


# Configure paths
MERGED_DIR = Path('../../data/merged')

# Create output directory if it doesn't exist
MERGED_DIR.mkdir(parents=True, exist_ok=True)

print(f"Output location: {MERGED_DIR.resolve()}")
print(f"Output directory created: {MERGED_DIR.exists()}")

# Load merged dataset
master = pd.read_csv(MERGED_DIR / 'merged.csv')
```

```python
print(f"Shape of df before feature engineering: {master.shape}")
print(f"Columns before feature engineering: {master.columns.tolist()}")
```

## 2. Ratio of sites and clicks (Feature Engineering)


'num_sites / num_unique_sites'


```python
# Calculate site engagement ratios
master['sites_revisit_ratio'] = master['num_sites'] / master['num_unique_sites'].replace(0, 1)
master['activity_diversity_ratio'] = master['num_unique_sites'] / master['total_unique_activities'].replace(0, 1)

print("New ratio features created:")
print(f"sites_revisit_ratio: num_sites / num_unique_sites")
print(f"activity_diversity_ratio: num_unique_sites / total_unique_activities")
print(f"\nSample values:")
print(master[['num_sites', 'num_unique_sites', 'total_unique_activities', 'sites_revisit_ratio', 'activity_diversity_ratio']].head(10))
```

```python
print(f"Shape after feature engineering: {master.shape}")
```

## 3. Remove 'num_sites', 'num_unique_sites', and 'total_unique_activities'


```python
# Columns to remove
repeated_cols = ['num_sites', 'num_unique_sites', 'total_unique_activities']

# Dropping repeated cols
master = master.drop(columns=repeated_cols, inplace=False)

# Shape after deleting repeated cols
print(f"Shape after deleting {repeated_cols}: {master.shape}")
print(f"\nRemaining columns: {master.columns.tolist()}")
```

## Save the master df


```python
print(f"\nFinal missing values:")
print(master.isnull().sum())
```

Saving the Final Merged file


```python
# Save updated master
MERGED_DIR.mkdir(parents=True, exist_ok=True)
master.to_csv(MERGED_DIR / "merged_1.csv", index=False)

print(f"\n✓ Updated master dataframe saved to {MERGED_DIR / 'merged_1.csv'}")
print(f"Final shape: {master.shape}")
```
